```python



def factorial(n):
    """Calcula n!"""

    if not isinstance(n, int):
        raise TypeError("n debe ser un entero")

    if n < 0:
        raise ValueError("n no puede ser negativo")

    resultado = 1

    for i in range(2, n + 1):
        resultado *= i

    return resultado



def per_circular(n):
    """(n-1)!"""

    if n < 1:
        raise ValueError("Debe existir al menos una persona")

    return factorial(n - 1)


def circular_dos_juntos(n):
    """
    Dos personas específicas deben quedar juntas

    Fórmula:
    2(n-2)!
    """

    if n < 2:
        raise ValueError(
            "Se necesitan al menos 2 personas"
        )

    return 2 * factorial(n - 2)


def circular_dos_separados(n):
    """
    Dos personas específicas NO pueden quedar juntas

    Fórmula:
    (n-1)! - 2(n-2)!
    """

    if n < 2:
        raise ValueError(
            "Se necesitan al menos 2 personas"
        )

    return per_circular(n) - circular_dos_juntos(n)


def lider_entre(n):
    """
    Un líder debe quedar entre dos personas específicas

    Fórmula:
    2(n-3)!
    """

    if n < 3:
        raise ValueError(
            "Se necesitan al menos 3 personas"
        )

    return 2 * factorial(n - 3)


def grupos_separados(n, grupos):
    """
    Aproximación simple.

    Considera los grupos como bloques.
    """

    if grupos < 2:
        raise ValueError(
            "Debe haber al menos dos grupos"
        )

    if grupos > n:
        raise ValueError(
            "No puede haber más grupos que personas"
        )

    return factorial(n - grupos)


while True:

    try:

        n = int(
            input("Ingrese el número de personas: ")
        )

        if n <= 0:
            print(
                "Ingrese un número entero positivo."
            )
            continue

        break

    except ValueError:
        print(
            "Error: debe ingresar un número entero."
        )



while True:

    print("\n===== MENÚ =====")
    print("1. Permutaciones circulares")
    print("2. Dos personas deben quedar juntas")
    print("3. Dos personas no pueden quedar juntas")
    print("4. Un líder entre dos personas específicas")
    print("5. Varios grupos deben sentarse separados")
    print("6. Mostrar explicación")
    print("0. Salir")

    opcion = input("Seleccione una opción: ")

    try:

        if opcion == "1":

            print(
                f"\nResultado: {per_circular(n)}"
            )

        elif opcion == "2":

            print(
                f"\nResultado: {circular_dos_juntos(n)}"
            )

        elif opcion == "3":

            print(
                f"\nResultado: {circular_dos_separados(n)}"
            )

        elif opcion == "4":

            print(
                f"\nResultado: {lider_entre(n)}"
            )

        elif opcion == "5":

            grupos = int(
                input(
                    "Ingrese la cantidad de grupos: "
                )
            )

            print(
                f"\nResultado: "
                f"{grupos_separados(n, grupos)}"
            )

        elif opcion == "6":

            print("""
SIMETRÍA POR ROTACIÓN

En una mesa circular no existe una silla inicial.

Si todas las personas se desplazan una posición,
la distribución sigue siendo la misma.

Por ello se fija una persona y se permutan
únicamente las restantes.

Número de permutaciones circulares:

(n - 1)!
""")

        elif opcion == "0":

            print(
                "\nGracias por utilizar el programa."
            )
            break

        else:

            print("Opción inválida.")

    except Exception as e:

        print(f"Error: {e}")
```
