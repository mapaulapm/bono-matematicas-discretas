```python

from math import comb



def validar_n(n):
    if not isinstance(n, int):
        raise ValueError("n debe ser entero")
    if n < 0:
        raise ValueError("n debe ser >= 0")


def validar_k(n, k):
    if not isinstance(k, int):
        raise ValueError("k debe ser entero")
    if k < 0 or k > n:
        raise ValueError("k debe estar entre 0 y n")




def total_cadenas(n):
    validar_n(n)
    return 2 ** n


def exactamente_k_unos(n, k):
    validar_n(n)
    validar_k(n, k)
    return comb(n, k)


def a_lo_mas_k_unos(n, k):
    validar_n(n)
    validar_k(n, k)
    return sum(comb(n, i) for i in range(0, k + 1))


def al_menos_k_unos(n, k):
    validar_n(n)
    validar_k(n, k)
    return sum(comb(n, i) for i in range(k, n + 1))


def igual_ceros_y_unos(n):
    validar_n(n)

    if n % 2 != 0:
        return "No es posible (n debe ser par)"

    return comb(n, n // 2)




def generar_cadenas(n):
    return [format(i, f"0{n}b") for i in range(2 ** n)]


def filtrar_por_unos(cadenas, k):
    return [c for c in cadenas if c.count("1") == k]




while True:

    print("\n===== MENÚ =====")
    print("1. Todas las cadenas (2^n)")
    print("2. Exactamente k unos")
    print("3. A lo sumo k unos")
    print("4. Al menos k unos")
    print("5. Igual número de ceros y unos")
    print("6. Listar cadenas (n ≤ 10)")
    print("0. Salir")

    opcion = input("Seleccione una opción: ")

    try:

        if opcion == "1":
            n = int(input("n: "))
            print("Resultado:", total_cadenas(n))

        elif opcion == "2":
            n = int(input("n: "))
            k = int(input("k: "))
            print("Resultado:", exactamente_k_unos(n, k))

        elif opcion == "3":
            n = int(input("n: "))
            k = int(input("k: "))
            print("Resultado:", a_lo_mas_k_unos(n, k))

        elif opcion == "4":
            n = int(input("n: "))
            k = int(input("k: "))
            print("Resultado:", al_menos_k_unos(n, k))

        elif opcion == "5":
            n = int(input("n: "))
            print("Resultado:", igual_ceros_y_unos(n))

        elif opcion == "6":
            n = int(input("n: "))

            if n > 10:
                print("n demasiado grande para listar cadenas")
            else:
                cadenas = generar_cadenas(n)
                for c in cadenas:
                    print(c)

        elif opcion == "0":
            print("Saliendo...")
            break

        else:
            print("Opción inválida")

    except Exception as e:
        print("Error:", e)
```
